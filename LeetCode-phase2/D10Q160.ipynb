```python
class Solution(object):
    def getIntersectionNode(self, headA, headB):
        """
        :type head1, head1: ListNode
        :rtype: ListNode
        """
        startA = headA
        A = 0
        while startA:
            startA = startA.next
            A += 1
        
        startB = headB
        B = 0
        while startB:
            startB = startB.next
            B += 1
        
        if A > B:
            diff = A - B
            while diff != 0:
                headA = headA.next
                diff -= 1

        if B > A:
            diff = B - A
            while diff != 0:
                headB = headB.next
                diff -= 1

        while headA:
            if headA is headB:
                return headA
            
            headA = headA.next
            headB = headB.next

        return None
```
