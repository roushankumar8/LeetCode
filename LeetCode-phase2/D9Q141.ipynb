### Leetcode Question no 141.(Linked List Cycle)

```python
class Solution(object):
    def hasCycle(self, head):
        slow = fast = head
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
            if slow == fast:
                return True
        return False
```

### This is final solution done using slow fast pointer.
