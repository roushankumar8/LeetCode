## Leetcode Question no 1290.(Convert Binary number in a Linked List to a integer)

```python
class Solution(object):
    def getDecimalValue(self, head):
        """
        :type head: Optional[ListNode]
        :rtype: int
        """
        result = 0
        while head:
            result = result * 2 + head.val
            head = head.next
        return result
```
